## Exercise 1 – The receipt (Easy)
Read an item, a quantity and a unit price. VAT is 18%. Print three lines.

Try: notebook, 3, 12.5. Then try 7 and 0.1 and look at the first line.

*Plan: 3 min solo · 5 min together.*

```python
item = input("Item: ")
quantity = int(input("Quantity: "))
price = float(input("Unit price: "))

subtotal = quantity * price
vat = subtotal * 0.18
total = subtotal + vat

print(f"{item} x {quantity} = {subtotal}")
print(f"VAT (18%): {round(vat, 2)}")
print(f"Total: {round(total, 2)}")
```

## Exercise 1.1 – A discount
Try: notebook, 3, 12.5, 10 → 33.75, 6.08, 39.83

*Plan: 0 min solo · 2 min together.*

```python
item = input("Item: ")
quantity = int(input("Quantity: "))
price = float(input("Unit price: "))
discount = float(input("Discount (%): "))
subtotal = quantity * price
paid = subtotal * (1 - discount / 100)
vat = paid * 0.18
total = paid + vat
print(f"{item} x {quantity} = {subtotal}")
print(f"After discount: {round(paid, 2)}")
print(f"VAT (18%): {round(vat, 2)}")
print(f"Total: {round(total, 2)}")
```

### The same solution, with readable values
A correct value isn't always easy to understand when it comes to time: 3600 is right, but 60 * 60 says what it is. Name the values once and use the names.

```python
total = int(input("Seconds: "))

seconds_per_minute = 60
seconds_per_hour = 60 * 60

hours = total // seconds_per_hour
rest = total % seconds_per_hour
minutes = rest // seconds_per_minute
seconds = rest % seconds_per_minute

print(f"{hours} hours")
print(f"{minutes} minutes")
print(f"{seconds} seconds")
```

### Palindrome
Try 121, 123, then 1221.

```python
n = int(input("A three-digit number: "))

first = n // 100
last = n % 10
print(f"Palindrome: {first == last}")
```

### What about other numbers?
Two digits: 40 prints True and 44 prints False – both wrong. More than three digits always prints False (n // 100 is at least 10, n % 10 at most 9) – wrong for 1221.

Guard without `if`: turn the promise into a value and combine it.

```python
n = int(input("A number: "))

valid = 100 <= n <= 999
print(f"Three digits: {valid}")
print(f"Palindrome: {valid and n // 100 == n % 10}")
```

```python
hours = int(input("Hours now: "))
minutes = int(input("Minutes now: "))
duration = int(input("Duration in minutes: "))

total = hours * 60 + minutes + duration
end_hours = (total // 60) % 24
end_minutes = total % 60

tens = end_minutes // 10
ones = end_minutes % 10
print(f"Ends at {end_hours}:{tens}{ones}")
```

## Extra 2 – Four digits
Try 4913.

*Plan: 2 min solo · 4 min together.*

```python
n = int(input("Number: "))

d1 = n // 1000          # thousands
d2 = n % 1000 // 100    # hundreds
d3 = n % 100 // 10      # tens
d4 = n % 10             # units
backwards = d4 * 1000 + d3 * 100 + d2 * 10 + d1

print(f"Digits: {d1} {d2} {d3} {d4}")
print(f"Sum: {d1 + d2 + d3 + d4}")
print(f"Backwards: {backwards}")
print(f"Palindrome: {n == backwards}")
```
