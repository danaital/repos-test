## H1 · Is it a double?
Roll two dice and print "Double!" if they show the same face, otherwise "Not a double". No if – use a bool times a string.

*6 minutes on your own.*

```python
# h1_double.py - is it a double? (no if)
import random

die1 = int(random.random() * 6) + 1
die2 = int(random.random() * 6) + 1

print(f"Dice: {die1} and {die2}")
print((die1 - die2 == 0) * "Double!" + (die1 - die2 != 0) * "Not a double")
```

## H2 · Booleans are numbers
Read three whole numbers a, b, c. Without if, abs, min or max, print: the sign of a (−1, 0 or 1); how many of the three are even; whether exactly one is positive; whether a is divisible by 3 or 5, but not both.

*8 minutes on your own.*

```python
# h2_booleans.py - booleans are numbers (no if, abs, min or max)
a = int(input("a: "))
b = int(input("b: "))
c = int(input("c: "))

sign_a = (a > 0) - (a < 0)
evens = (a % 2 == 0) + (b % 2 == 0) + (c % 2 == 0)
exactly_one_positive = (a > 0) + (b > 0) + (c > 0) == 1
fizz_xor_buzz = (a % 3 == 0) != (a % 5 == 0)

print(f"Sign of a: {sign_a}")
print(f"Even numbers: {evens}")
print(f"Exactly one positive: {exactly_one_positive}")
print(f"a divisible by 3 or 5, not both: {fizz_xor_buzz}")
```

## H3 · Across midnight
A shift starts and ends on a 24-hour clock. It's shorter than a day but may cross midnight. Print its length as h:mm. (22:30 → 1:15 gives 2:45.)

*10 minutes on your own.*

```python
# h3_midnight.py - the length of a shift that may cross midnight
start_h = int(input("Start hour: "))
start_m = int(input("Start minute: "))
end_h = int(input("End hour: "))
end_m = int(input("End minute: "))

diff = (end_h * 60 + end_m - (start_h * 60 + start_m)) % 1440
hours = diff // 60
minutes = diff % 60
print(f"Duration: {hours}:{minutes // 10}{minutes % 10}")
```

## H5 · A loaded die
A 6 comes up half the time; each of 1–5 comes up 10% of the time. One call to random.random(), no if.

*10 minutes on your own.*

```python
# h5_loaded_die.py - a 6 half the time, each of 1-5 10% of the time
import random

r = random.random()
six = int(r >= 0.5)                 # 1 half the time, else 0
other = int(r * 10) + 1             # 1..5 when r < 0.5
face = six * 6 + (1 - six) * other
print(f"The loaded die shows {face}")
```
